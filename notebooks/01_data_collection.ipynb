```python
date_params = {
    "select": "min(date) as min_date, max(date) as max_date"
}

date_response = requests.get(
    url,
    params=date_params,
    timeout=30
)

date_response.status_code
```

```python
date_data = date_response.json()
date_data["results"]
```

```python
date_range = date_data["results"][0]
date_range
```

```python
min_date = pd.Timestamp(date_range["min_date"]).tz_convert(None)
max_date = pd.Timestamp(date_range["max_date"]).tz_convert(None)

months = pd.period_range(
    min_date.to_period("M"),
    max_date.to_period("M"),
    freq="M"
)

months
```

```python
monthly_counts = []

for month in months:
    start_date = month.start_time.strftime("%Y-%m-%d")
    next_month = (month + 1).start_time.strftime("%Y-%m-%d")

    params = {
        "where": (
            f"date >= date'{start_date}' "
            f"AND date < date'{next_month}'"
        ),
        "limit": 1
    }

    response_month = requests.get(
        url,
        params=params,
        timeout=30
    )

    response_month.raise_for_status()

    monthly_counts.append({
        "month": str(month),
        "records": response_month.json()["total_count"]
    })
    
```

```python
monthly_counts_df = pd.DataFrame(monthly_counts)

monthly_counts_df
```

```python
export_url = (
    "https://opendata.tpg.ch/api/explore/v2.1/catalog/datasets/"
    "montees-par-arret-par-ligne/exports/csv"
)
```

```python
def download_month(month):
    period = pd.Period(month, freq="M")

    start_date = period.start_time.strftime("%Y-%m-%d")
    next_month = (period + 1).start_time.strftime("%Y-%m-%d")

    params = {
        "where": (
            f"date >= date'{start_date}' "
            f"AND date < date'{next_month}'"
        ),
        "delimiter": ","
    }

    response = requests.get(
        export_url,
        params=params,
        timeout=120
    )

    response.raise_for_status()

    return response
```
